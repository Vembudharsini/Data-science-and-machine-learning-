```python
import pandas as pd
```

```python
df = pd.read_csv("googleplaystore.csv")
df
```

```python
df.columns[df.isna().any()]
```

```python
df.isnull().sum()
```

```python
df = df.drop(df.index[10472])
df
```

```python
df = df.drop(["App", "Last Updated", "Current Ver", "Android Ver"], axis=1, errors='ignore')
df
```

```python
df.columns[df.isna().any()]
```

```python
df.isnull().sum()
```

```python
df['Rating'] = df['Rating'].fillna(df['Rating'].median())
df['Is_Successful'] = (df['Rating'] >= 4.2).astype(int)
df
df['Size'] = df['Size'].fillna(df['Size'].median())

# Check if there are any NaNs left
print("Missing values in Size:", df['Size'].isnull().sum())
df
```

```python
df['Installs'] = df['Installs'].astype(str).str.replace('+', '', regex=False).str.replace(',', '', regex=False)
df['Installs'] = pd.to_numeric(df['Installs'], errors='coerce').fillna(0).astype(int)
df['Price'] = df['Price'].astype(str).str.replace('$', '', regex=False)
df['Price'] = pd.to_numeric(df['Price'], errors='coerce').fillna(0).astype(float)
df
```

```python
import numpy as np
def clean_size(x):
    # Convert to string, lowercase, and remove spaces
    x = str(x).lower().strip() 
    
    if 'm' in x:
        # Extract only the numbers/decimals and convert to float
        return float(''.join(c for c in x if c.isdigit() or c == '.'))
    elif 'k' in x:
        # Extract digits, convert to float, then convert KB to MB
        val = float(''.join(c for c in x if c.isdigit() or c == '.'))
        return val / 1024
    else:
        # This catches "varies with device" or any other text
        return np.nan

df['Size'] = df['Size'].apply(clean_size)
```

```python
fo = df.select_dtypes(include="object").columns
for i in fo:
    df[i] = df[i].fillna(df[i].mode()[0])
df
    
```

```python
df['Size'] = df['Size'].fillna(df['Size'].median())

# Check again—this should now be 0
print("Missing values in Size:", df['Size'].isnull().sum())
df
```

```python
from sklearn.preprocessing import LabelEncoder 
le = LabelEncoder()
for i in fo:
    df[i] = le.fit_transform(df[i])
df
```

```python
x = df.drop(["Is_Successful", "Rating"], axis=1)
y = df["Is_Successful"]
```

```python
df['Reviews'] = pd.to_numeric(df['Reviews'], errors='coerce').fillna(0)
df
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.ensemble import RandomForestClassifier
model = RandomForestClassifier(n_estimators=300, max_depth=20, random_state=42, class_weight="balanced")
model.fit(x_train, y_train)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
y_pred = model.predict(x_test)
print("Accuracy Score:", accuracy_score(y_test, y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```
